```python
# 5. Countries with internet penetration greater than 70%

# Re-establishing merged_df and calculating 'Internet Penetration (%)
merged_df = population.merge(gdp, on='country', how='inner')
merged_df = merged_df.merge(Jsonn, on='country', how='inner')
merged_df = merged_df.merge(xmllll, on='country', how='inner')

merged_df['Internet Penetration (%)'] = (merged_df['internet_users'] / merged_df['population']) * 100

merged_df[merged_df['Internet Penetration (%)'] > 70][['country', 'Internet Penetration (%)']]
```

```python
# 6. Top 5 countries by GDP per capita
# Re-establishing merged_df and calculating 'GDP Per Capita'
merged_df = population.merge(gdp, on='country', how='inner')
merged_df = merged_df.merge(Jsonn, on='country', how='inner')
merged_df = merged_df.merge(xmllll, on='country', how='inner')

merged_df['GDP Per Capita'] = merged_df['GDP'] / merged_df['population']

merged_df.nlargest(5, 'GDP Per Capita')[['country', 'GDP Per Capita']]
```

```python
# 13. Display all countries having literacy rate above 90%:
q = '''SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
ORDER BY literacy_rate DESC '''
df = pd.read_sql_query(q, conn)
df
```

```python
# 14. Find the Top 5 countries with the highest internet penetration:

q = ''' SELECT country, \"Internet Penetration (%)\"
FROM country_statistics
ORDER BY \"Internet Penetration (%)\" DESC LIMIT 5 '''
df = pd.read_sql_query(q, conn)
df
```
