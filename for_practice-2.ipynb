```python
#Word Frequency CounterGiven a paragraph of text, write a function that returns the 3 mostfrequently occurring words
 #(ignore case and punctuation).


def word_count(text):
    words = text.lower().split()
    count = {}

    for word in words:
        word = word.strip(".,!?")
        count[word] = count.get(word, 0) + 1

    return sorted(count, key=count.get, reverse=True)[:3]


text = input("Enter text: ")
print(word_count(text))

```

```python
# Password ValidatorWrite a function that checks if a password is strong: at least 8 characters,
#contains at least one digit, one uppercase letter, and one special character(!@#$%).




def check_password(password):
    if len(password) < 8:
        return False

    digit = False
    uppercase = False
    special = False

    for ch in password:
        if ch.isdigit():
            digit = True
        elif ch.isupper():
            uppercase = True
        elif ch in "!@#$%":
            special = True

    return digit and uppercase and special

password= input("Enter password: ")

if check_password(password):
    print("Strong password")
else:
    print("Weak password")
```

```python
import math


numbers = [10, 12, 15, 18, 20, 20, 22, 25, 30, 100]


# Q1. Range
def find_range(numbers):
    return min(numbers), max(numbers)


# Q2. Mean
def mean(numbers):
    return sum(numbers) / len(numbers)


# Q3. Median
def median(numbers):
    numbers = sorted(numbers)
    n = len(numbers)

    if n % 2 == 1:
        return numbers[n // 2]
    else:
        return (numbers[n // 2 - 1] + numbers[n // 2]) / 2


# Q4. Mode
def mode(numbers):
    count = {}

    for n in numbers:
        count[n] = count.get(n, 0) + 1

    highest = max(count.values())

    if highest == 1:
        return None

    result = [n for n in count if count[n] == highest]

    if len(result) == 1:
        return result[0]
    else:
        return result


# Q5. Percentile
def percentile(numbers, p):
    numbers = sorted(numbers)

    index = (p / 100) * (len(numbers) - 1)

    lower = int(index)
    upper = lower + 1

    if upper >= len(numbers):
        return numbers[lower]

    return numbers[lower] + (index - lower) * (
        numbers[upper] - numbers[lower]
    )


# Q6. Lower bound
def lower_bound(numbers):
    q1 = percentile(numbers, 25)
    q3 = percentile(numbers, 75)

    iqr = q3 - q1

    return q1 - 1.5 * iqr


# Q7. Outliers
def outliers(numbers):
    q1 = percentile(numbers, 25)
    q3 = percentile(numbers, 75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    result = [n for n in numbers if n < lower or n > upper]

    if len(result) == 0:
        return None

    return result


# Q8. Standard deviation
def standard_deviation(numbers):
    avg = mean(numbers)

    total = 0

    for n in numbers:
        total += (n - avg) ** 2

    variance = total / len(numbers)

    return math.sqrt(variance)


# Q9. Z-score
def z_scores(numbers):
    avg = mean(numbers)
    sd = standard_deviation(numbers)

    result = []

    for n in numbers:
        z = (n - avg) / sd
        result.append(z)

    return result


# Q10. Pearson correlation
def pearson(x, y):
    x_mean = mean(x)
    y_mean = mean(y)

    numerator = 0
    x_total = 0
    y_total = 0

    for i in range(len(x)):
        numerator += (x[i] - x_mean) * (y[i] - y_mean)
        x_total += (x[i] - x_mean) ** 2
        y_total += (y[i] - y_mean) ** 2

    return numerator / math.sqrt(x_total * y_total)


# Display outputs

print("Q1 Range:", find_range(numbers))

print("Q2 Mean:", mean(numbers))

print("Q3 Median:", median(numbers))

print("Q4 Mode:", mode(numbers))

print("Q5 90th Percentile:", percentile(numbers, 90))

print("Q6 Lower Bound:", lower_bound(numbers))

print("Q7 Outliers:", outliers(numbers))

print("Q8 Standard Deviation:", standard_deviation(numbers))

print("Q9 Z-Scores:", z_scores(numbers))


x = [10, 20, 30, 40, 50]
y = [20, 40, 60, 80, 100]

print("Q10 Pearson Correlation:", pearson(x, y))
```
