```python
#Word Frequency CounterGiven a paragraph of text, write a function that returns the 3 mostfrequently occurring words
 #(ignore case and punctuation).


def word_count(text):
    words = text.lower().split()
    count = {}

    for word in words:
        word = word.strip(".,!?")
        count[word] = count.get(word, 0) + 1

    return sorted(count, key=count.get, reverse=True)[:3]


text = input("Enter text: ")
print(word_count(text))

```

```python
# Password ValidatorWrite a function that checks if a password is strong: at least 8 characters,
#contains at least one digit, one uppercase letter, and one special character(!@#$%).




def check_password(password):
    if len(password) < 8:
        return False

    digit = False
    uppercase = False
    special = False

    for ch in password:
        if ch.isdigit():
            digit = True
        elif ch.isupper():
            uppercase = True
        elif ch in "!@#$%":
            special = True

    return digit and uppercase and special

password= input("Enter password: ")

if check_password(password):
    print("Strong password")
else:
    print("Weak password")
```
