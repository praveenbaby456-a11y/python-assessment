```python
#5
movie_data = movie_data.rename(columns={"Title":"Movie_Title"})
```

```python
#6
movie_data["Genre"].unique()
```

```python
#7

movie_data['Num_Genres'] = movie_data['Genre'].str.count(',') + 1
movie_data[['Movie_Title','Num_Genres']]
```

```python
#8
reve_mean = movie_data['Revenue (Millions)'].mean()
metascore_mean = movie_data['Metascore'].mean()
movie_data['Revenue (Millions)'] = movie_data['Revenue (Millions)'].fillna(reve_mean)
movie_data['Metascore'] = movie_data['Metascore'].fillna(metascore_mean)
movie_data.isnull().sum()
```

```python
#9
movie_data.nlargest(5, 'Rating')[['Movie_Title', 'Rating']]
```

```python
#10
highest_rev = movie_data.loc[movie_data['Revenue (Millions)'].idxmax()]
lowest_rev = movie_data.loc[movie_data['Revenue (Millions)'].idxmin()]
print("highest revevnue",highest_rev[['Movie_Title', 'Revenue (Millions)']])
print("lowest revevnue",lowest_rev[['Movie_Title', 'Revenue (Millions)']])
```

```python
#11
movie_data.groupby('Director')['Rating'].mean().sort_values(ascending=False)
```

```python
#12
movie_data['Year'].value_counts().sort_index()
```

```python
#13
high_rated_high_revenue = movie_data[(movie_data['Rating'] > 8) & (movie_data['Revenue (Millions)'] > 100)]
high_rated_high_revenue[['Movie_Title', 'Rating', 'Revenue (Millions)']]
```

```python
#14
movie_data.groupby('Director').agg(
    {'Rating': 'mean', 'Revenue (Millions)': 'mean'}
    ).sort_values(by='Rating', ascending=False)
```

```python
#15
movie_data.groupby('Year').agg(
    {'Movie_Title': 'count', 'Rating': 'mean'}
    ).rename(columns={'Movie_Title': 'Movie_Count'})
```

```python
#16
movie_data['Profitability'] = movie_data['Revenue (Millions)'] / movie_data['Runtime (Minutes)']
movie_data.nlargest(10, 'Profitability')[['Movie_Title', 'Profitability']]
```

```python
#17
def rating_group(rating):
    if rating >= 7.5:
        return 'Excellent'
    elif rating >= 6.0:
        return 'Average'
    else:
        return 'Poor'

movie_data['Rating_Category'] = movie_data['Rating'].apply(rating_group)
movie_data[['Movie_Title', 'Rating', 'Rating_Category']]
```

```python
#18
def calculate_score(row):
    return (row['Rating'] * 10) + row['Revenue (Millions)']

movie_data['Score'] = movie_data.apply(calculate_score, axis=1)
movie_data[['Movie_Title', 'Rating', 'Revenue (Millions)', 'Score']].sort_values(by='Score', ascending=False)
```
